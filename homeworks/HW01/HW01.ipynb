```python
# 1. Импорт библиотеки NumPy
import numpy as np
```

```python
# 2. Примеры создания массивов
examples = {
    'a': np.array([1, 2, 3]),  # Создание массива
    'b': np.array([[1, 2, 3], [4, 5, 6]]),  # Создание двумерного массива (матрица)
    'c': np.zeros((3, 4)),  # Создание матрицы из 0 с размерами 3 на 4
    'd': np.ones((2, 3, 4)),  # Создание трехмерного массива из 1 с параметрами: 2 матрицы 3 на 4
    'e': np.eye(4),  # создание единичной матрицы
    'g': np.empty((2, 2)),  # Создание пустой матрицы 2 на 2
    'g': np.arange(20, 30, 5)  # Создание массива от 20(включительно) до 30(не включитаельно) с шагом 5
     
}

for example in examples:
    print(f'{example}) {examples[example]}')
```

```python
# 3. Демонстрация атрибутов shape и dtype
print(f'shape a): {examples['a'].shape}')
print(f'dtype a): {examples['a'].dtype}')
```

```python
# 4. Примеры индексирования и срезов
first = examples['a']
second = examples['b']
print('1) Получение отдельного элемента', second[1, 2])
print('2) Получение отдельного элемента', first[2])
print('3) Получение отдельных элементов', first[[1, 2]])
print('4) Получение отдельных элементов', second[1, [0,2]]) 
print('5) Получение подмассива', first[:2])
print('6) Получение подмассива', first[1:2])
print('7) Получение обратного подмассива', first[::-1])
```

```python
# 5. Примеры векторных операций и базовых статистик
print('1) Поэлементное сложение двух матриц:', second + first)
print('2) Поэлементное умножение двух матриц:', first * second)
print('3) Поэлементное вычитание двух матриц:', first - second)
print('4) Поэлементное сложение с числом:', second + 1)
print('5) Вычисление средней величины массива:', first.mean())
print('6) Вычисление средней величины по столбцам:', second.mean(axis=0))
print('7) Вычисление средней величины по строкам:', second.mean(axis=1))

```
